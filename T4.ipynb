```python
from pylab import *

# Cargar datos
data = loadtxt("/content/stars.dat")
temp = data[:,0]
mag  = data[:,1]

# Graficar
figure(figsize=(7,5))
scatter(temp, mag, s=10)

xlabel("Temperature")
ylabel("Magnitude")

# Invertir ejes (como en HR real)
gca().invert_xaxis()
gca().invert_yaxis()

title("HR Diagram (pylab)")
grid(True)
show()
```

```python
import numpy as np

# Cargar datos A (omitir líneas con '#')
A = []
with open("/content/altura5.dat") as f:
    for line in f:
        if line.startswith("#") or line.strip()=="":
            continue
        A.append(float(line.split()[0]))
A = np.array(A)

figure(figsize=(7,5))
hist(A, bins=20)
xlabel("Altura A (m)")
ylabel("Frecuencia")
title("Histograma de alturas A (pylab)")
grid(True)
show()
```
